<h4>Q3. Write a program to implement AND & OR gates using Perceptron</h4>

```python


def perceptron_AND(x1, x2):
    w1 = 1
    w2 = 1
    bias = -2
    net = w1*x1 + w2*x2 + bias
    if net >= 0:
        return 1
    else:
        return 0

print("AND Gate")
inputs = [(0,0), (0,1), (1,0), (1,1)]
for x1, x2 in inputs:
    print(f"{x1} AND {x2} = {perceptron_AND(x1, x2)}")

print("\n")

def perceptron_OR(x1, x2):
    w1 = 1
    w2 = 1
    bias = -1
    net = w1*x1 + w2*x2 + bias
    if net >= 0:
        return 1
    else:
        return 0

print("OR Gate")
inputs = [(0,0), (0,1), (1,0), (1,1)]
for x1, x2 in inputs:
    print(f"{x1} OR {x2} = {perceptron_OR(x1, x2)}")
```

<h4>Q4. Implementation of XOR Problem using PyTorch Neural Network</h4>

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python

A = torch.tensor([[0., 0.],
                  [0., 1.],
                  [1., 0.],
                  [1., 1.]])

B = torch.tensor([[0.],
                  [1.],
                  [1.],
                  [0.]])
```

```python
class XORNet(nn.Module):
    def __init__(self):
        super(XORNet, self).__init__()
        self.hidden = nn.Linear(2, 4)  
        self.output = nn.Linear(4, 1)

    def forward(self, x):
        x = torch.relu(self.hidden(x)) 
        x = self.output(x)             
        return x
```

```python
model = XORNet()

criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

epochs = 5000
for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(A)
    loss = criterion(outputs, B)
    loss.backward()
    optimizer.step()
```

```python
epochs = 10000

for epoch in range(epochs):
    optimizer.zero_grad()
    
    outputs = model(A)
    loss = criterion(outputs, B)
    
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 1000 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
with torch.no_grad():
    logits = model(A)
    predictions = torch.sigmoid(logits)
    predictions = (predictions > 0.5).int()
    print(predictions)
```

<h4>Q5. Implement Simple below Neural Network to solve regression problem</h4>

```python

A = torch.tensor([[10., 12.],
                  [32., 20.],
                  [3., 9.],
                  [14., 18.]], dtype=torch.float32)

B = torch.tensor([[2.],
                  [9.],
                  [25.],
                  [7.]], dtype=torch.float32)
```

```python
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()

        self.hidden = nn.Linear(2, 2)
        

        self.output = nn.Linear(2, 1)
        
    def forward(self, x):
        h = torch.relu(self.hidden(x))   
        y = self.output(h)               
        return y
```

```python
model = SimpleNN()

criterion = nn.MSELoss()    
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
epochs = 5000

for epoch in range(epochs):
    optimizer.zero_grad()
    
    predictions = model(A)
    loss = criterion(predictions, B)
    
    loss.backward()
    optimizer.step()
    
    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

```python
with torch.no_grad():
    test_input = torch.tensor([[23., 68.]])
    output = model(test_input)
    print("Predicted output:", output.item())
```
